```python
import sys; sys.path.insert(0, '..')
import pandas as pd
risk = pd.read_parquet('../data/processed/lad_risk_table.parquet')
risk.sort_values('overheating_risk_score', ascending=False).head(15)
```

## Priority filter: high risk × high private-rented share
These LADs combine high exposure with low household agency.

```python
priority = risk[(risk['risk_tier'].isin(['High','Severe'])) & (risk['share_rented_private'] > 0.30)]
priority = priority.sort_values('overheating_risk_score', ascending=False)
priority[['lad_name','overheating_risk_score','risk_tier','share_flats','share_rented_private']]
```

## Three policy levers this analysis supports

1. **Retrofit existing stock.** Approved Document O regulates only *new* builds. The risk concentration here is in existing dense urban housing.
2. **Renter protections.** Decent Homes 2025 should include overheating thresholds as well as cold-home minima.
3. **Cool-roof / urban-green programmes.** Highest-leverage where flat share AND urban density both peak.
